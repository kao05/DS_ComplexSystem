```python
#@title Divine Comedy - English
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve
        local, _ = urlretrieve(url, filename)
        print('Downloaded ' + local)
```

```python
download('https://www.gutenberg.org/cache/epub/8799/pg8799.txt')
```

```python
first_line = "CANTO I\n"   # English
last_line ='            *** END OF THE PROJECT GUTENBERG EBOOK THE DIVINE COMEDY BY DANTE, ILLUSTRATED, PARADISE, COMPLETE ***\n'
file_name = 'pg8799.txt'   # English
fp = open(file_name)

for line in fp:
    if line == first_line:
        break

unique_words = {}
for line in fp:
    if line == last_line:
        break

    for word in line.split():
          if word in unique_words:
              unique_words[word] += 1
          else:
              unique_words[word] = 1
```

```python
print('Total words:',len(unique_words))
dic_word = list(unique_words)
freq_word = list(unique_words.values())

print(dic_word[:10])
print(dic_word[-10:])
```

Se podrá observar que hay información irrelevante en el texto, por lo que es necesario hacer una limpieza. Los pasos usuales incluyen:

Filtrado: stop-words como artículos, conjunciones, preposiciones, etc. Palabras extremadamente raras.

Lematización: verbos en infinitivo, minúsculas, sustantivos en singular.

"Stemming": formas básicas de las palabras. Porter (1980) propuso un conocido algoritmo de codificación basado en reglas.

```python
import sklearn
```

```python
docs = ['Mirabai has won a silver medal in weight lifting in Tokyo olympics 2021',
                 'Sindhu has won a bronze medal in badminton in Tokyo olympics',
                 'Indian hockey team is in top four team in Tokyo olympics 2021 after 40 years']
```

Además, observemos la instanciación de la clase, métodos y propiedades.

```python
from sklearn.feature_extraction.text import CountVectorizer

# La clase que hará la bolsa de palabras
vectorizer = CountVectorizer()

# Construimos la bolsa de palabras
bag = vectorizer.fit_transform(docs)
# print(type(bag))

# Imprimimos las palabras del vocabulario
print(f"Palabras del vocabulario:\n{vectorizer.get_feature_names_out()}")

# Un diccionario con el índice de cada palabra
print(f"\nDiccionario con el índice de cada palabra:\n{vectorizer.vocabulary_}")

# La matriz de features
matriz = bag.toarray()
print(f"\nDataset con esta representación:\n{matriz}")
print(f"\nShape de la matriz {matriz.shape}")
```
