```python
#@title Divine Comedy - English
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve
        local, _ = urlretrieve(url, filename)
        print('Downloaded ' + local)
```

```python
download('https://www.gutenberg.org/cache/epub/8799/pg8799.txt')
```

```python
first_line = "CANTO I\n"   # English
last_line ='            *** END OF THE PROJECT GUTENBERG EBOOK THE DIVINE COMEDY BY DANTE, ILLUSTRATED, PARADISE, COMPLETE ***\n'
file_name = 'pg8799.txt'   # English
fp = open(file_name)

for line in fp:
    if line == first_line:
        break

unique_words = {}
for line in fp:
    if line == last_line:
        break

    for word in line.split():
          if word in unique_words:
              unique_words[word] += 1
          else:
              unique_words[word] = 1
```

```python
print('Total words:',len(unique_words))
dic_word = list(unique_words)
freq_word = list(unique_words.values())

print(dic_word[:10])
print(dic_word[-10:])
```

Se podrá observar que hay información irrelevante en el texto, por lo que es necesario hacer una limpieza. Los pasos usuales incluyen:

Filtrado: stop-words como artículos, conjunciones, preposiciones, etc. Palabras extremadamente raras.

Lematización: verbos en infinitivo, minúsculas, sustantivos en singular.

"Stemming": formas básicas de las palabras. Porter (1980) propuso un conocido algoritmo de codificación basado en reglas.

```python
#@title Importación del dataset y librerias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LinearRegression

# 1. Cargar el dataset de Forbes
df = pd.read_csv('DS_ComplexSystem/data/forbes_richman.csv', encoding='latin-1')

# 2. Crear el corpus concatenando las columnas textuales
corpus = (df['Name'].fillna('') + ' ' +
          df['Country'].fillna('') + ' ' +
          df['Source'].fillna('') + ' ' +
          df['Industry'].fillna('')).tolist()

df.head()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LinearRegression

# Instanciar CountVectorizer (sin filtrado inicial de stopwords)
vectorizer = CountVectorizer()
bag = vectorizer.fit_transform(corpus)

# Obtener vocabulario y frecuencias globales sumando por columnas (axis=0)
words = vectorizer.get_feature_names_out()
counts = bag.toarray().sum(axis=0)

# Crear DataFrame ordenado por frecuencia y asignar Rango
df_zipf = pd.DataFrame({'word': words, 'frequency': counts})
df_zipf = df_zipf.sort_values(by='frequency', ascending=False).reset_index(drop=True)
df_zipf['rank'] = df_zipf.index + 1

# --- Gráficos Quest 1 ---
# 1. Ajustar el tamaño general de la figura (más ancha)
plt.figure(figsize=(20, 6))

# Subplot 1: Primeras 60 palabras
plt.subplot(1, 2, 1)
plt.bar(df_zipf['word'][:60], df_zipf['frequency'][:60])
plt.xticks(rotation=90, fontsize=8)  # Se reduce el tamaño del texto para dar espacio
plt.title("Primeras 60 palabras (Rango vs Frecuencia)")

# Subplot 2: Últimas 60 palabras
plt.subplot(1, 2, 2)
plt.bar(df_zipf['word'][-60:], df_zipf['frequency'][-60:])
plt.xticks(rotation=90, fontsize=8)  # Se reduce el tamaño del texto para dar espacio
plt.title("Últimas 60 palabras (Rango vs Frecuencia)")

plt.tight_layout()
plt.show()

# 3. Comprobar Ley de Zipf mediante Regresión Lineal en escala Log-Log
df_zipf['log_rank'] = np.log10(df_zipf['rank'])
df_zipf['log_freq'] = np.log10(df_zipf['frequency'])

X = df_zipf[['log_rank']]
y = df_zipf['log_freq']

reg = LinearRegression().fit(X, y)
pendiente = reg.coef_[0]
interseccion = reg.intercept_

plt.figure(figsize=(7, 5))
plt.scatter(df_zipf['log_rank'], df_zipf['log_freq'], alpha=0.5, label='Datos')
plt.plot(df_zipf['log_rank'], reg.predict(X), color='red', label=f'Ajuste: y = {pendiente:.2f}x + {interseccion:.2f}')
plt.xlabel(r'$\log_{10}(\text{Rango})$')
plt.ylabel(r'$\log_{10}(\text{Frecuencia})$')
plt.title("Ley de Zipf - Escala Logarítmica")
plt.legend()
plt.show()

print(f"Pendiente de la ley de potencias: {pendiente:.4f}")
```

```python
# Mismo procedimiento con filtrado de stop-words
vectorizer_clean = CountVectorizer(stop_words='english')
bag_clean = vectorizer_clean.fit_transform(corpus)

words_clean = vectorizer_clean.get_feature_names_out()
counts_clean = bag_clean.toarray().sum(axis=0)

df_zipf_clean = pd.DataFrame({'word': words_clean, 'frequency': counts_clean})
df_zipf_clean = df_zipf_clean.sort_values(by='frequency', ascending=False).reset_index(drop=True)
df_zipf_clean['rank'] = df_zipf_clean.index + 1
```

```python
def calcular_entropia(frecuencias):
    probabilidades = frecuencias / frecuencias.sum()
    # Filtrar probabilidades mayor a 0 para evitar log2(0)
    probabilidades = probabilidades[probabilidades > 0]
    return -np.sum(probabilidades * np.log2(probabilidades))

entropia_original = calcular_entropia(df_zipf['frequency'].values)
entropia_filtrada = calcular_entropia(df_zipf_clean['frequency'].values)

print(f"Entropía con stopwords: {entropia_original:.4f} bits")
print(f"Entropía sin stopwords: {entropia_filtrada:.4f} bits")
```
